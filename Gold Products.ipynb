### dlt pipeline

streaming Table


```python
import dlt
from pyspark.sql.functions import *
```

```python
my_rules = {
    "rule1": "product_id IS NOT NULL",
    "rule2": "product_name IS NOT NULL"
}
```

```python
@dlt.table(
    name="DimProducts_stage"
)
@dlt.expect_all(my_rules)
def dimproducts_stage():
    return (
        spark.readStream
        .table("databricks_cat.silver.products_silver")
    )
```

Streaming View

```python
@dlt.view(
    name="DimProducts_view"
)
def dimproducts_view():
    return dlt.read_stream("DimProducts_stage")
```

**Dim Products**

```python
dlt.create_streaming_table(
    name="DimProducts"
)
```

```python
dlt.apply_changes(
    target="DimProducts",
    source="DimProducts_view",
    keys=["product_id"],
    sequence_by=col("product_id"),   # MUST be timestamp/version
    stored_as_scd_type=2
)
```
