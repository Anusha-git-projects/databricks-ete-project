```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

## Data REading

```python
df= spark.read.format("parquet")\
  .load("abfss://bronze@databricksprojectanushaa.dfs.core.windows.net/customers")
```

```python
df.display()
```

```python
df=df.withColumn("domaims",split(col('email'),'@')[1])
df.display()
```

```python
df= df.withColumn("full_name",concat(col('first_name'),lit(' '),col('last_name')))
df=df.drop('first_name','last_name')
df.display()
```

```python
df.write.mode("overwrite").format("delta").save("abfss://silver@databricksprojectanushaa.dfs.core.windows.net/customers")
```
